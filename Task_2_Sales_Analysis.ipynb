```python
category_sales = (
    sales_data.groupby("CategoryName")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

product_sales = (
    sales_data.groupby("ProductName")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

print("Revenue by Category")
print(category_sales.to_string(index=False))

print("\nTop 10 Products by Revenue")
print(product_sales.head(10).to_string(index=False))

print("\nTop Category:", category_sales.iloc[0]["CategoryName"])
print("Best Product:", product_sales.iloc[0]["ProductName"])
```

```python
sales_data["Month"] = sales_data["OrderDate"].dt.to_period("M").astype(str)

monthly_sales = (
    sales_data.groupby("Month")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Month")
)

print(monthly_sales.to_string(index=False))
```
